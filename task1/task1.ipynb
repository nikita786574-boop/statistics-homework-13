```python
import numpy as np
```

```python
x = np.array([7, 5, 10])
y = np.array([30, 30, 42])

matrix_covariation  = np.cov(x, y)
print(matrix_covariation)
print(matrix_covariation[0,1])
print(np.var(x, ddof = 1))
b1 = matrix_covariation[0,1] / np.var(x, ddof=1)
```

```python
print(round(b1, 4))
b0 = y.mean() - b1 * x.mean()
print(round(b0,4))
```
